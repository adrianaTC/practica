```python
%%writefile Fabrica.java
import java.util.ArrayList;
public class Fabrica {
    private ArrayList<Sucursal> sucursales;

    public Fabrica() {
        this.sucursales = new ArrayList<>();
    }

    public void agregarSucursal(Sucursal sucursal) {
        sucursales.add(sucursal);
    }

    public void listarinstrumentos() {
        for (Sucursal s : sucursales) {
            s.listarinstrumentos();
        }
    }

    public ArrayList<Instrumento> instrumentosPorTipo(Tipoinstrumento tipo) {
        ArrayList<Instrumento> acumulado = new ArrayList<>();
        for (Sucursal s : sucursales) {
            acumulado.addAll(s.instrumentosPorTipo(tipo));
        }
        return acumulado;
    }

    public Instrumento borrarinstrumento(String ID) {
        for (Sucursal s : sucursales) {
            Instrumento borrado = s.borrarinstrumento(ID);
            if (borrado != null) return borrado;
        }
        return null;
    }

    public double[] porcinstrumentosPorTipo(String nombreSucursal) {
        Sucursal s = buscarSucursal(nombreSucursal);
        if (s != null) {
            return s.porcinstrumentosPorTipo();
        }
        return new double[3]; //3 pq son percusion, viento y cuerda
    }

    private Sucursal buscarSucursal(String nombre) {
        for (Sucursal s : sucursales) {
            if (s.getNombre().equals(nombre)) return s;
        }
        return null;
    }
}

```
